# **Tourism Analysis at The Amazon Monument in Benin**  

### **Background**  
The **Amazon Monument** in Benin, inaugurated on **July 30, 2022**, honors the legendary Amazons of the Kingdom of Dahomey. Similar to the **Monument de la Renaissance Africaine** in Dakar, it celebrates **African heritage, pride, resilience**, and symbolizes the **strength and historical legacy** of the people.

Since its inauguration, the monument has become a popular destination, attracting visitors from all over the world. To monitor tourism trends, the **Agence Nationale du Tourisme Béninois** has started recording data **every three days** on various aspects of visitor activity at the monument.  It is stored in the `data` folder as `Tourist_Visits_at_Amazon_Monument.csv`

Your task is to **analyze the collected data**  to uncover trends and patterns that can help the tourism agency improve visitor experiences and manage operations better.  


### **Dataset Overview**

The dataset provided contains the following columns:

1. **Date**:  
   - The date when the data was recorded (every 3 days).  

2. **Number of Adults**:  
   - The total number of adult visitors (18 years and above).  

3. **Number of Teenagers**:  
   - The total number of teenage visitors (13-17 years old).  

4. **Number of Children**:  
   - The total number of child visitors (under 13 years).  

5. **Local vs Foreign Visitors Ratio**:  
   - The proportion of local (Beninese) visitors compared to foreign tourists (value between 0 and 1).  


6. **Daily Revenue (USD)**:  
   - The total revenue generated from ticket sales and other activities on that day.  

7. **Weather Conditions**:  
   - The weather on the recorded day (Sunny, Rainy, Cloudy).  

8. **Temperature (°C)**:  
   - The average temperature on the recorded day.  
 

10. **Photo-Taking Frequency**:  
   - The number of photos taken or social media check-ins on that day, reflecting visitor engagement.

### **6. Beginning of June until end of August is often the time period where tourists are flowing into the country. Provide a comparative plot of the number of visits for each age group (children, teenagers, adults) during that time period in 2023? Which age group  seems to be the  most represented?**

```python
# Step 1 : Make sur "date" is datatime pandas
# Step 2 : Filter juin-august 2023
# Step 3 : Calcul of total visitors by age group
# Step 4 : Create graph
# Step 5 : Personnalizing graph
```

```python
# 1. Charger et préparer les données
amazon_visits_df['date'] = pd.to_datetime(amazon_visits_df['date']) 

# Conversion de la colonne 'date' en datetime

# 2. Filtrer les données entre juin et août 2023
start_date = pd.Timestamp('2023-06-01')
end_date = pd.Timestamp('2023-08-31')

summer_amazon_visits_df = amazon_visits_df[
    (amazon_visits_df['date'] >= start_date) 
    & (amazon_visits_df['date'] <= end_date)]
```

```python
plt.figure(figsize=(8, 6))
bars = plt.bar(age_groups, visit_counts, color=['skyblue', 'orange', 'green'])

# Ajouter les valeurs au-dessus des barres
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 50, int(yval), ha='center', va='bottom')

# 5. Personnaliser le graphique
plt.title('Number of Visits by Age Group (June–August 2023)')
plt.ylabel('Total Visits')
plt.xlabel('Age Group')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### **7. Display the average monthly revenues for each season in 2023. (you could use the `groupby` function). Which month generated the most revenues and which season does it map to?**


```python
# Step 1 : Filtering by 2023 year
# Step 2 : Extract specific months
# Step 3 : Define function for sorted season in Benin
# Step 4 : Calculate monthly and mean revenue
# Step 5 : Monthly visualisation and season
```

```python
# 1. Filter only 2023
amazon_visits_df_2023 = amazon_visits_df[amazon_visits_df['year'] == 2023].copy()
# 2. Extraire le mois
amazon_visits_df_2023['month'] = amazon_visits_df_2023['date'].dt.month
```

```python
# 3. Let's define a function for a season

def get_benin_season(month):
    if month in [4, 5, 6, 7]:
        return 'Gde saison des pluies'
    elif month in [8, 9]:
        return 'Pte saison sèche'
    elif month in [10, 11]:
        return 'Pte saison des pluies'
    else:  # Décembre à Mars
        return 'Gde saison sèche'
   
amazon_visits_df_2023['season'] = amazon_visits_df_2023['month'].apply(get_benin_season)
```

```python
# 4. Calculer les revenus moyens par mois
monthly_avg = amazon_visits_df_2023.groupby('month')['daily_revenue_usd'].mean()

# 5. Calculer les revenus moyens par saison
season_avg = amazon_visits_df_2023.groupby('season')['daily_revenue_usd'].mean()

season_avg
```

```python
# 6. Identify month with highest revenue
max_month = monthly_avg.idxmax()
max_value = monthly_avg.max()
max_season = get_benin_season(max_month)
```

```python
# 7. Monthly vizualisation revenues
plt.figure(figsize=(10,6))
monthly_avg.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Monthly Revenues in 2023')
plt.xlabel('Month')
plt.ylabel('Average Revenue (USD)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 8. Season vizualisation revenues
plt.figure(figsize=(8,5))
season_avg.plot(kind='bar', color='orange', edgecolor='black')
plt.title('Average Seasonal Revenues in 2023')
plt.xlabel('Season')
plt.ylabel('Average Revenue (USD)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Print max month with max value and max season 
print(f"Le mois {max_month} a généré le plus de revenus ({max_value}),  et correspond à la saison : {max_season}.")
```
