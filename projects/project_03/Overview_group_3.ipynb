## 📊 Dataset Description

You have been provided access to the council's mining database in a CSV file named `south_african_mining_data.csv` containing comprehensive operational data from mines across South Africa.

### Data Fields:

**Identification & Location:**
- **Mine_ID**: Unique mine identifier
- **Mine_Name**: Name of the mining operation
- **Province**: South African province (Gauteng, Northern Cape, Eastern Cape, etc.)
- **Primary_Mineral**: Main mineral extracted (Gold, Platinum, Diamonds, Coal, Manganese, Iron Ore, etc.)
- **Mine_Type**: Type of mining operation (Underground, Open-pit, Alluvial)
- **Ownership_Type**: Ownership structure (Private, State-Owned, Joint Venture)

**Operational Metrics:**
- **Annual_Production_Tons**: Annual production volume in metric tons
- **Employees**: Number of employees
- **Years_Operating**: How long the mine has been operational
- **Production_per_Employee**: Productivity metric (tons per employee)

**Safety Metrics:**
- **Safety_Incidents_2023**: Number of safety incidents in 2023
- **Safety_Rate_per_1000_Employees**: Safety incident rate normalized per 1000 employees

**Financial Metrics:**
- **Revenue_Million_ZAR**: Annual revenue in millions of South African Rand
- **Operating_Costs_Million_ZAR**: Annual operating costs in millions of ZAR
- **Profit_Million_ZAR**: Annual profit in millions of ZAR
- **Profit_Margin_Percent**: Profit margin percentage

**Sustainability Metrics:**
- **Environmental_Compliance_Score**: Environmental compliance score (0-100)
- **Community_Investment_Million_ZAR**: Investment in local communities in millions of ZAR
- **Water_Usage_Million_Liters**: Annual water consumption in millions of liters
- **Energy_Consumption_MWh**: Annual energy consumption in megawatt-hours
