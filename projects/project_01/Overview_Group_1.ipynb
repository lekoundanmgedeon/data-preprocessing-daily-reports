# Project 01: Customer Analytics for NigeriaBank

## 🏦 Business Context

**NigeriaBank** is one of Nigeria's leading commercial banks with over 500 branches across the country, serving millions of customers from Lagos to Kano, from Rivers to Delta State. The bank has been experiencing challenges in understanding customer behavior, optimizing branch performance, and managing loan portfolios effectively. These challenges are affecting the bank's profitability and customer satisfaction levels.

The executive management team has identified several critical business questions:
- **Customer Segmentation**: Which customer segments are most profitable? How do demographics influence banking behavior?
- **Digital Transformation**: What is the adoption rate of digital banking across different regions and age groups?
- **Loan Portfolio Management**: What patterns exist in loan uptake and repayment? Which customer profiles are most creditworthy?
- **Regional Performance**: How do banking patterns vary across Nigerian states?
- **Customer Retention**: What factors influence customer longevity with the bank?

As a newly hired **Data Analyst** at NigeriaBank, you have been tasked with analyzing the bank's customer database to uncover actionable insights that can enhance customer relationship management, improve branch operations, and optimize lending decisions.

---

## 📊 Dataset Description

You have been provided access to the bank's customer database in a CSV file named `nigerian_banking_data.csv` containing comprehensive customer information.

### Data Fields:
- **Customer_ID**: Unique customer identifier
- **Customer_Name**: Customer name
- **Age**: Customer age
- **Gender**: Male/Female
- **State**: Nigerian state where customer is based
- **Account_Type**: Savings or Current account
- **Account_Balance_NGN**: Current account balance in Nigerian Naira
- **Monthly_Income_NGN**: Customer's monthly income in Naira
- **Employment_Status**: Employed, Self-Employed, Retired, Student, Unemployed
- **Years_With_Bank**: Duration of relationship with the bank
- **Has_Loan**: Whether customer has an active loan (True/False)
- **Loan_Type**: Type of loan if applicable
- **Loan_Amount_NGN**: Loan amount in Naira
- **Credit_Score**: Customer credit score (300-850)
- **Monthly_Transactions**: Average number of monthly transactions
- **Digital_Banking_User**: Whether customer uses digital banking (True/False)
- **Account_Opening_Date**: Date when account was opened

---

## 🎯 Your Mission

Your line manager is particularly interested in understanding customer demographics, account types, transaction patterns, and loan performance, as this information is crucial for making informed decisions on marketing strategies, branch optimization, and risk management.

### Expected Deliverables:

1. **Exploratory Data Analysis (EDA)**
   - Perform comprehensive data exploration to understand the structure, quality, and characteristics of the dataset
   - Identify and handle missing values, outliers, and data quality issues
   - Generate summary statistics for key variables

2. **Customer Segmentation Analysis**
   - Analyze customer demographics (age, gender, state, employment status)
   - Examine account types and balance distributions
   - Identify high-value customer segments

3. **Digital Banking Adoption**
   - Analyze digital banking adoption rates across different customer segments
   - Identify factors that influence digital banking usage
   - Provide recommendations for increasing digital adoption

4. **Loan Portfolio Analysis**
   - Examine loan uptake patterns across customer segments
   - Analyze relationship between credit scores, income, and loan amounts
   - Identify risk factors and opportunities in lending

5. **Regional and Temporal Patterns**
   - Analyze banking patterns across Nigerian states
   - Examine customer acquisition trends over time
   - Identify seasonal or temporal patterns in customer behavior

6. **Data Visualization**
   - Create compelling visualizations to communicate your findings
   - Use appropriate charts and graphs for different types of insights
   - Ensure visualizations are clear, informative, and business-focused

7. **Actionable Insights and Recommendations**
   - Provide data-driven recommendations for marketing strategies
   - Suggest improvements for customer retention and acquisition
   - Recommend risk management strategies for the loan portfolio
